# Variables


Variables are containers for storing data values.



We have seen Python has no command for declaring a variable. A variable is created the moment you first assign a value to it.

```python
x = 5
y = "John"
print(x)
print(y)
```

Variables do not need to be declared with any particular type, and can even change type after they have been set.

```python
x = 4       # x is of type int
x = "Sally" # x is now of type str
print(x)
```

**Casting**:
If you want to specify the data type of a variable, this can be done with casting.

```python
x = str(3) # X will be '3'
y = int(3) # Y will be 3
z = float(3) # z will be 3.0
print(x)
print(y)
print(z)
```

**Get the Type:** You can get the data types of a variable with type() function.

```python
print(type(x))
print(type(y))
print(type(z))
```

**Quotation in Python**: Python accepts single ('), double (") and triple (''' or """) quotes to denote string literals, as long as the same type of quote starts and ends the string.

```python
x = "John"
# is the same as
x = 'John'
x = '''John'''
x = """John"""


print(x)
```
